Q14.
text = "    Data Analysis with Python is very powerful    "

Perform the following tasks:

Remove extra spaces using strip().
Convert the string into title case using title().
Replace "Powerful" with "Important".
Find the position of the word "Python".

```python
text = " Data Analysis with Python is very powerful "
strip=text.strip()
title=text.title()
replace=text.replace("powerful","Important")
find=text.find("pyhton")
print("Remove extra spaces:                 ",strip)
print("Convert the string into title case:  ",title)
print("Replace 'Powerful' with 'Important': ",replace)
print("the position of the word 'Python' is:",find)
```
