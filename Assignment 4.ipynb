Q1 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]
print(users)
for i in range(len(users)):
    if i == 3:
        print("Limit reached, Stop emails.")
        break
    else:
        print("Email sent to :",users[i])

        
```

Q2. Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] critical = 40

Ouput:

Temperature: 22

Temperature: 25

Temperature: 30

Critical temperature reached! Stop monitoring.

```python
temps = [22, 25, 30, 45, 28] 
critical = 40
print(temps)
for j in temps:
    if j >= critical:
        print("Critical temperature reached! Stop monitoring")
        break
    else:
        print("Temperature: ",j)
        
```

Q3. Calculate total price of items in a shopping cart (Using For loop)

cart = [100, 250, 50, 300]

Output:

Total Price: 700

```python
cart = [100, 250, 50, 300]
total=0
for k in cart:
    total=total+k
print("Total Price: ",total)
```

Q4. Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True}
]

Output:

Access granted to: Ali

Access granted to: John

```python
customers = [ {"name": "Ali", "paid": True},{"name": "Sara", "paid": False},{"name": "John", "paid": True}]
print(customers)
for a in customers:
    if a["paid"] == False:
        continue
    else:
        print("Access granted to: ",a["name"])
```

Q6. Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

Output:

domains: ['gmail.com', 'yahoo.com', 'hotmail.com']



```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]
print(emails)
for email in emails:
    domain=email.split("@")[1]
    print(domain)
```

Q8. Create a multiplication table using a nested for loop.

numbers = [2, 3, 4]
multipliers = [1, 2, 3]

Expected Output:

2 x 1 = 2
2 x 2 = 4
2 x 3 = 6
3 x 1 = 3
3 x 2 = 6
3 x 3 = 9
4 x 1 = 4
4 x 2 = 8
4 x 3 = 12

Use a nested for loop.

```python
numbers = [2, 3, 4] 
multipliers = [1, 2, 3]

for num in numbers:
    for mul in multipliers:
        print(num,"x",mul,"=",num*mul)
```

Q9. Keep asking the user for a password until the correct password is entered.

correct_password = "python123"

Expected behaviour:

Enter password: abc
Incorrect password. Try again.

Enter password: 12345
Incorrect password. Try again.

Enter password: python123
Access granted!

Use a while loop.

```python
user_input=input("enter the correct password")
correct_password = "python123"
while user_input != correct_password:
    user_input  
    print("Incorrect password, Try again")
    user_input=input("enter the correct password")
print("Accessed granted!")
```

Q10. Count down from 5 to 1 and then display "Time's up!".

Expected Output:

5
4
3
2
1
Time's up!

Use a while loop.

```python
nums=5
while nums>=1:
    print(nums,end=" ")
    nums-=1
print("time`s up!")
```
